```python
def detect_outliers_iqr(data, feature):
    Q1 = data[feature].quantile(0.25)
    Q3 = data[feature].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return lower_bound, upper_bound

def detect_outliers_zscore(data, feature, threshold=3):
    z_scores = np.abs((data[feature] - data[feature].mean()) / data[feature].std())
    return z_scores > threshold
```

```python
# Analyze TransactionAmt outliers
if 'TransactionAmt' in train_data.columns:
    # IQR Method
    lower_iqr, upper_iqr = detect_outliers_iqr(train_data, 'TransactionAmt')
    outliers_iqr = train_data[(train_data['TransactionAmt'] < lower_iqr) | 
                             (train_data['TransactionAmt'] > upper_iqr)]
    
    # Z-score Method
    outliers_zscore = train_data[detect_outliers_zscore(train_data, 'TransactionAmt')]
    
    print("Transaction Amount Outlier Analysis:")
    print(f"IQR Outliers: {len(outliers_iqr):,} transactions ({len(outliers_iqr)/len(train_data)*100:.2f}%)")
    print(f"Z-score Outliers: {len(outliers_zscore):,} transactions ({len(outliers_zscore)/len(train_data)*100:.2f}%)")
    
    if 'isFraud' in train_data.columns:
        print(f"\nFraud Rate in IQR Outliers: {outliers_iqr['isFraud'].mean()*100:.2f}%")
        print(f"Fraud Rate in Z-score Outliers: {outliers_zscore['isFraud'].mean()*100:.2f}%")
        
        normal_range = train_data[(train_data['TransactionAmt'] >= lower_iqr) & 
                                 (train_data['TransactionAmt'] <= upper_iqr)]
        print(f"Fraud Rate in Normal Range: {normal_range['isFraud'].mean()*100:.2f}%")
    
    # Visualize outliers
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 3, 1)
    plt.hist(train_data['TransactionAmt'], bins=50, edgecolor='black', alpha=0.7)
    plt.axvline(upper_iqr, color='red', linestyle='--', label=f'IQR Upper: ${upper_iqr:.2f}')
    plt.title('Transaction Amount Distribution with IQR Bound')
    plt.xlabel('Transaction Amount')
    plt.ylabel('Frequency')
    plt.yscale('log')
    plt.legend()
    
    plt.subplot(1, 3, 2)
    # Box plot
    plt.boxplot(train_data['TransactionAmt'].dropna())
    plt.title('Transaction Amount Box Plot')
    plt.ylabel('Amount')
    
    plt.subplot(1, 3, 3)
    if 'isFraud' in train_data.columns:
        fraud_amt = train_data[train_data['isFraud'] == 1]['TransactionAmt']
        non_fraud_amt = train_data[train_data['isFraud'] == 0]['TransactionAmt']
        sns.kdeplot(non_fraud_amt, label='Non-Fraud', fill=True, alpha=0.6)
        sns.kdeplot(fraud_amt, label='Fraud', fill=True, alpha=0.6)
        plt.title('Transaction Amount by Fraud Status')
        plt.xlabel('Transaction Amount')
        plt.legend()
        plt.xscale('log')
    
    plt.tight_layout()
    plt.show()
```
